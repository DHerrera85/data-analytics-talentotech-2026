## 1. Ejemplo de visualización con Matplotlib

```python

import pandas as pd
import matplotlib.pyplot as plt

# Crear un DataFrame de ejemplo
data = {
    'Producto': ['A', 'B', 'C', 'D'],
    'Ventas': [150, 200, 300, 100]
}

df = pd.DataFrame(data)
```

```python

# Crear un gráfico de barras
plt.bar(df['Producto'], df['Ventas'], color='skyblue')
plt.title('Ventas por Producto')
plt.xlabel('Producto')
plt.ylabel('Ventas')
plt.show()
```

## 1.1. Ejemplo de visualización con Seaborn

```python

import seaborn as sns

# Crear un gráfico de distribución de ventas
sns.histplot(df['Ventas'], bins=4, kde=True)
plt.title('Distribución de Ventas')
plt.xlabel('Ventas')
plt.ylabel('Frecuencia')
plt.show()
```

## 2. Ejemplo de un EDA: Base de datos «Tips» de Seaborn

### A. Cargar la base de datos

```python

import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

# Cargar la base de datos tips de Seaborn
tips = sns.load_dataset("tips")

# Visualizar las primeras filas del DataFrame
print(tips.head())
```

### B. Resumen estadístico de los datos

```python

# Obtener un resumen estadístico de tips
print(tips.describe())
```

**Prácticas complementarias:** las tres visualizaciones siguientes refuerzan la lectura de distribuciones, relaciones y categorías; no corresponden a las actividades finales del PDF.

### Práctica complementaria — Histograma de cuentas

```python
# Visualizar la distribución de las cuentas
sns.histplot(tips['total_bill'], bins=20, kde=True)

plt.title('Distribución de las Cuentas')
plt.xlabel('Importe de la cuenta')
plt.ylabel('Frecuencia')
plt.show()
```

### Práctica complementaria — Cuenta y propina (dispersión simple)

```python

# Relación entre el importe de la cuenta y la propina
sns.scatterplot(data=tips, x='total_bill', y='tip')

plt.title('Relación entre Cuenta y Propina')
plt.xlabel('Importe de la cuenta')
plt.ylabel('Propina')
plt.show()
```

### Práctica complementaria — Cuentas por día

```python

# Cantidad de cuentas registradas por día
sns.countplot(data=tips, x='day')

plt.title('Cantidad de Cuentas por Día')
plt.xlabel('Día de la semana')
plt.ylabel('Cantidad de cuentas')
plt.show()
```

### C. Análisis de distribución — Histogramas

```python

# Visualizar la distribución de las propinas
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.histplot(tips['tip'], bins=30, kde=True)
plt.title('Distribución de Propinas')
plt.xlabel('Propina')
plt.ylabel('Frecuencia')

# Visualizar la distribución del total de la cuenta
plt.subplot(1, 2, 2)
sns.histplot(tips['total_bill'], bins=30, kde=True)
plt.title('Distribución del Total de la Cuenta')
plt.xlabel('Total de la Cuenta')
plt.ylabel('Frecuencia')

plt.tight_layout()
plt.show()
```

### C. Análisis de distribución — Diagramas de caja

```python

# Visualizar la distribución de las propinas
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
sns.boxplot(tips['tip'])
plt.title('Distribución de Propinas')

# Visualizar la distribución del total de la cuenta
plt.subplot(1, 2, 2)
sns.boxplot(tips['total_bill'])
plt.title('Distribución del Total de la Cuenta')

plt.show()
```

### D. Análisis de la relación entre variables

```python

plt.figure(figsize=(8, 6))
sns.scatterplot(data=tips, x='total_bill', y='tip',
                hue='day', style='sex', s=100)
plt.title('Relación entre Total de la Cuenta y Propina')
plt.xlabel('Total de la Cuenta')
plt.ylabel('Propina')
plt.show()
```

## Actividad 1: Análisis de ventas de diamantes

```python

# Actividad 1 - Análisis de Ventas de Diamantes

# Cargar el conjunto de datos diamonds
diamonds = sns.load_dataset("diamonds")
```

```python

# Visualizar las primeras filas del DataFrame
print(diamonds.head())
```

```python

# Actividad 1 - Punto 3
# Resumen estadístico de las variables numéricas

print(diamonds.describe())
```

```python

# Actividad 1 - Punto 4
# Distribución de los precios de los diamantes

plt.figure(figsize=(10, 5))
sns.histplot(diamonds['price'], bins=30, kde=True)

plt.title('Distribución de Precios de Diamantes')
plt.xlabel('Precio')
plt.ylabel('Frecuencia')
plt.show()
```

```python

# Actividad 1 - Punto 5
# Distribución de diamantes por color

plt.figure(figsize=(10, 5))
sns.countplot(data=diamonds, x='color')

plt.title('Distribución de Diamantes por Color')
plt.xlabel('Categoría de Color')
plt.ylabel('Cantidad de Diamantes')
plt.show()
```

### Actividad 1 — Conclusiones

Se analizaron **53.940 diamantes**, cuyo precio promedio fue de **3.932,80**, mientras que el precio máximo registrado alcanzó los **18.823**.

El análisis de distribución permitió observar que la mayor cantidad de diamantes se concentra en los precios bajos, mientras que los valores más elevados presentan una menor frecuencia.

Respecto de la clasificación por color, la categoría **G** registró la mayor cantidad de diamantes, mientras que la categoría **J** fue la menos frecuente.

Para analizar los precios se utilizó un **histograma**, debido a que permite observar la distribución de una variable numérica. En cambio, para comparar las categorías de color se empleó un **gráfico de barras**, apropiado para representar las frecuencias de una variable categórica.

## Actividad 2: Análisis de supervivencia

```python

# Actividad 2 - Análisis de Supervivencia

# Cargar la base de datos Titanic
titanic = sns.load_dataset("titanic")

# Visualizar las primeras filas
print(titanic.head())
```

```python

# Actividad 2 - Punto 2
# Calcular la tasa de supervivencia por clase

supervivencia_por_clase = titanic.groupby('pclass')['survived'].mean() * 100

print(supervivencia_por_clase)
```

```python

# Actividad 2 - Punto 3
# Gráfico de barras de supervivencia por clase

plt.figure(figsize=(8, 5))
sns.barplot(x=supervivencia_por_clase.index,
            y=supervivencia_por_clase.values)

plt.title('Tasa de Supervivencia por Clase')
plt.xlabel('Clase del Pasajero')
plt.ylabel('Tasa de Supervivencia (%)')
plt.ylim(0, 100)
plt.show()
```

```python

# Actividad 2 - Punto 4
# Relación entre edad y supervivencia

plt.figure(figsize=(8, 5))
sns.boxplot(data=titanic, x='survived', y='age')

plt.title('Distribución de Edades según Supervivencia')
plt.xlabel('Supervivencia (0 = No, 1 = Sí)')
plt.ylabel('Edad')
plt.show()
```

### Actividad 2 — Conclusiones

Se analizó la supervivencia de los pasajeros del Titanic utilizando el conjunto de datos de Seaborn.

El análisis permitió identificar diferencias importantes según la clase del pasajero. La primera clase presentó la mayor tasa de supervivencia, con un **62,96 %**, seguida por la segunda clase, con un **47,28 %**, y la tercera, con un **24,24 %**.

El gráfico de barras permitió visualizar estas diferencias, mientras que el diagrama de caja mostró que la mediana de edad fue aproximadamente igual entre quienes sobrevivieron y quienes no sobrevivieron, cercana a los **28 años**.

Los resultados muestran una asociación entre la clase del pasajero y la supervivencia. En cambio, el análisis de edades no permite establecer por sí solo una relación concluyente entre la edad y las posibilidades de sobrevivir.

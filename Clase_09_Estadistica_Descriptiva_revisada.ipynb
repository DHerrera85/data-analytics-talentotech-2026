Prácticas de NumPy y SciPy y actividades de ventas mensuales.

```python
import numpy as np
```

# Cuartiles y percentiles

```python
ventas = [100, 150, 200, 250, 300, 350, 400, 450, 500, 1000]

p75 = np.percentile(ventas, 75)

print(p75)
```

```python
ventas = [100, 150, 200, 250, 300, 350, 400, 450, 500, 1000]

p25 = np.percentile(ventas, 25)

print(p25)
```

```python
p50 = np.percentile(ventas, 50)

print(p50)
```

# Rango intercuartílico

```python
p75 = np.percentile(ventas, 75)
p25 = np.percentile(ventas, 25)

rango_intercuartilico = p75 - p25
print(rango_intercuartilico)
```

# Rango

```python
rango = max(ventas) - min(ventas)

print(rango)
```

# Varianza

```python
varianza = np.var(ventas)

print(varianza)
```

# Desviación estándar

```python
desviacion_estandar = np.std(ventas)

print(f"La desviación estándar es: {desviacion_estandar:.2f}")
```

# Actividad 1: Análisis de Tendencia Central

```python
ventas_mensuales = [1.2, 2.5, 3.1, 18.3, 40.5, 52.1,
                   54.8, 46.2, 25.5, 13.8, 11.9, 9.2]
```

```python
# 1 y 2. Calcular la media y la mediana

media_ventas = np.mean(ventas_mensuales)
mediana = np.median(ventas_mensuales)

print(f"La media es: {media_ventas:.2f} millones")
print(f"La mediana es: {mediana:.2f} millones")
```

```python
# 3. Ejemplo para comprender la moda (datos distintos de las ventas mensuales)
from scipy import stats

ventas_ejemplo = [10, 15, 15, 20, 25]
moda = stats.mode(ventas_ejemplo)[0]
print(f"La moda del ejemplo es: {moda}")

# En las 12 ventas mensuales originales ningún valor se repite:
# no hay una moda única.
```

# Actividad 2: Análisis de dispersión

```python
# 1. Calcular el rango de las ventas

rango_mensual = max(ventas_mensuales) - min(ventas_mensuales)
print(f"El rango mensual es: {rango_mensual:.2f} millones")
```

```python
# 2. Calcular la varianza de las ventas

varianza = np.var(ventas_mensuales)
print(f"La varianza es: {varianza:.2f} (millones)²")
```

```python
# 3. Calcular la desviación estándar de las ventas

desviacion_estandar = np.std(ventas_mensuales)
print(f"La desviación estándar es: {desviacion_estandar:.2f} millones")

```

## Interpretación de la Actividad 2

Las ventas mensuales muestran una dispersión considerable (rango: **53,60 millones**; desviación estándar: **19,19 millones**). El máximo se registra en **julio (54,8 millones)** y las ventas son elevadas durante el invierno argentino. Entre las opciones propuestas, el producto más compatible con este comportamiento estacional es la **bufanda**. Esta interpretación es una hipótesis basada en el patrón de ventas, no una identificación demostrada del producto.

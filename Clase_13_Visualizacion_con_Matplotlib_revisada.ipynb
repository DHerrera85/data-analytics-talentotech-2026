# Gráficos de Líneas

```python

import matplotlib.pyplot as plt

# Datos
tiempo = [1, 2, 3, 4, 5]
ventas = [10, 15, 20, 25, 30]

# Gráfico de líneas
plt.plot(tiempo, ventas)
plt.show()
```

# Personalización del Gráfico de Líneas

```python

plt.plot(tiempo, ventas, marker='s', linestyle='--',
         color='green', linewidth=2, markersize=10)
plt.title('Ventas a lo largo del tiempo', fontsize=14)
plt.xlabel('Años', fontsize=12)
plt.ylabel('Ventas', fontsize=12)
plt.grid(True)
plt.xticks(tiempo)
plt.show()
```

# Gráficos de Barras

```python

# Datos
categorias = ['A', 'B', 'C', 'D']
valores = [5, 7, 3, 8]

# Gráfico de barras
plt.bar(categorias, valores)
plt.show()
```

# Personalización del Gráfico de Barras

```python

plt.bar(categorias, valores, color=['red', 'blue', 'green',
'purple'], width=0.5)
plt.title('Comparación de categorías', fontsize=14)
plt.xlabel('Categorías', fontsize=12)
plt.ylabel('Valores', fontsize=12)
plt.xticks(rotation=45)
plt.show()
```

# Gráficos de Dispersión

```python

x = [1, 2, 3, 4, 5]
y = [2, 3, 5, 1, 4]

# Gráfico de dispersión
plt.scatter(x, y)
plt.show()
```

# Personalización del Gráfico de Dispersión

```python

plt.scatter(x, y, color='cyan', s=[100, 150, 200, 250, 300])
plt.title('Gráfico de dispersión de variables', fontsize=14)
plt.xlabel('Variable X', fontsize=12)
plt.ylabel('Variable Y', fontsize=12)
plt.grid(True)
plt.show()
```

# Actividad 1: Análisis de Ventas a lo Largo del Tiempo


```python

# Datos de ventas mensuales (en unidades)
meses = ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun', 'Jul',
         'Ago', 'Sep', 'Oct', 'Nov', 'Dic'] # Lista de los meses del año

ventas = [100, 120, 130, 150, 180, 170, 200, 210, 250, 270,
          300, 320] # Lista de las unidades vendidas correspondientes a cada mes
```

```python

# Gráfico de líneas de ventas mensuales
plt.plot(meses, ventas)
plt.title('Ventas mensuales')
plt.xlabel('Meses')
plt.ylabel('Unidades vendidas')
plt.show()
```

El gráfico muestra una tendencia general de crecimiento en las ventas mensuales del producto, que pasan de 100 unidades en enero a 320 en diciembre, lo que representa un incremento del 220 %. Aunque en junio se observa una leve disminución respecto de mayo, las ventas retoman su crecimiento durante los meses siguientes. Diciembre registra el mayor volumen de ventas del año, lo que evidencia una evolución positiva del producto durante el período analizado.

# Actividad 2: Comparación de Productos en una Campaña

```python

# Datos de ventas por producto
productos = ['Producto A', 'Producto B', 'Producto C',
             'Producto D'] # Lista de los nombres de los productos
ventas = [300, 250, 400, 350] # Lista de las unidades vendidas de cada producto en la campaña.
```

```python

# Gráfico de ventas por artículo
plt.bar(productos, ventas)
plt.show()
```

```python

# Personalización del gráfico de ventas por producto
colores = ['steelblue', 'steelblue', 'orange', 'steelblue']

plt.bar(productos, ventas, color=colores)
plt.title('Ventas por producto durante la campaña publicitaria')
plt.xlabel('Productos')
plt.ylabel('Unidades vendidas')
plt.show()
```

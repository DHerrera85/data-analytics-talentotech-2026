## 1. Introducción a la correlación

La **correlación** es una medida estadística que describe la relación entre dos variables. Permite identificar si los cambios de una variable tienden a estar acompañados por cambios en otra.

### Tipos de correlación

- **Positiva:** cuando una variable aumenta, la otra también tiende a aumentar.
- **Negativa:** cuando una variable aumenta, la otra tiende a disminuir.
- **Nula:** no se observa una relación lineal entre las variables.

### Coeficiente de correlación de Pearson

El coeficiente de Pearson (`r`) mide la fuerza y dirección de la relación lineal entre dos variables numéricas.

- **r = +1:** correlación positiva perfecta.
- **r = −1:** correlación negativa perfecta.
- **r = 0:** ausencia de correlación lineal.
- **Valores próximos a +1 o −1:** indican una relación lineal más fuerte.

**Importante:** correlación no implica causalidad. Una relación estadística entre dos variables no demuestra que una sea la causa de la otra.

## 2. Cálculo y visualización de correlaciones

### Matriz de correlación con Pandas

El método `DataFrame.corr()` permite calcular la correlación entre las variables numéricas de un DataFrame y devuelve una **matriz de correlación**.

En esta matriz:

- Los valores próximos a **+1** indican una relación lineal positiva fuerte.
- Los valores próximos a **−1** indican una relación lineal negativa fuerte.
- Los valores próximos a **0** indican una relación lineal débil o inexistente.

### Visualización con Seaborn

Utilizaremos dos herramientas principales:

- **`sns.heatmap()`**: representa gráficamente la matriz de correlación mediante colores y valores numéricos.
- **`sns.scatterplot()`**: permite observar la relación entre dos variables mediante un diagrama de dispersión.

Estas visualizaciones facilitan la identificación e interpretación de patrones entre variables.

*Fuente: Talento Tech 2026 — Clase 11, página 5.*

```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python

# Cargar el set de datos
df = sns.load_dataset("iris")
df
```

```python

# Exploramos la columna de especies
df['species'].unique()
```

```python

# Seleccionamos sólo los datos de una especie
setosa = df[df['species'] == 'setosa']

setosa
```

```python

# Obtenemos las variables numéricas
setosa = setosa.drop(['species'], axis='columns')

setosa
```

```python

# Crear la matriz de correlación
correlation_matrix = setosa.corr()
print(correlation_matrix)
```

```python

# Investigar la correlación entre el largo y el ancho de sépalo
correlation_matrix['sepal_length']['sepal_width']
```

```python

# Gráfico de dispersión: longitud y ancho del sépalo
sns.scatterplot(x='sepal_length', y='sepal_width', data=setosa)
plt.show()
```

```python

# Crear el mapa de calor
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
plt.show()
```

### Conclusiones del análisis de correlación — Iris (Setosa)

Se analizaron 50 observaciones de la especie *Setosa*, considerando cuatro variables numéricas: longitud y ancho del sépalo, y longitud y ancho del pétalo.

La matriz de correlación de Pearson permitió identificar que:

- La **mayor correlación** se encuentra entre la longitud y el ancho del sépalo (`r = 0,7425`), indicando una asociación lineal positiva fuerte.
- La **menor correlación** se observa entre el ancho del sépalo y la longitud del pétalo (`r = 0,1777`), indicando una asociación lineal positiva muy débil.
- Todas las correlaciones calculadas entre variables diferentes son positivas.
- El gráfico de dispersión permite observar la tendencia ascendente entre la longitud y el ancho del sépalo.
- El mapa de calor facilita la comparación visual de los coeficientes de correlación.

**Conclusión general:** el análisis evidencia que algunas características físicas de las flores Setosa presentan relaciones lineales más fuertes que otras. Sin embargo, estas asociaciones estadísticas no demuestran relaciones de causalidad.

## 3. Actividad 1: Análisis de Correlación

**Contexto:** actividad práctica de SynthData, propuesta en la Clase 11 de Talento Tech 2026.

### Objetivo

Calcular e interpretar la correlación entre el precio (`price`) y el peso en quilates (`carat`) de los diamantes que cumplen simultáneamente estas condiciones:

- Color: **F**.
- Corte: **Premium**.

### Procedimiento

1. Cargar el dataset `diamonds`.
2. Filtrar los diamantes de color F y corte Premium.
3. Calcular el coeficiente de correlación de Pearson entre `price` y `carat`.
4. Representar la relación mediante un gráfico de dispersión.
5. Interpretar los resultados y elaborar una conclusión.

*Fuente: Talento Tech 2026 — Clase 11, páginas 9 y 10. Se utiliza `diamonds` en lugar de `iris` para resolver el enunciado.*

```python

# Cargar el dataset de diamantes
diamonds = sns.load_dataset("diamonds")

# Visualizar el DataFrame
diamonds
```

```python

# Filtrar diamantes de color F y corte Premium
diamonds_filtrados = diamonds[
    (diamonds['color'] == 'F') &
    (diamonds['cut'] == 'Premium')
]

# Visualizar los datos filtrados
diamonds_filtrados
```

```python

# Calcular la correlación entre quilates y precio
correlacion = diamonds_filtrados['carat'].corr(
    diamonds_filtrados['price']
)

print("Correlación entre quilates y precio:", correlacion)
```

```python

# Gráfico de dispersión entre quilates y precio
sns.scatterplot(
    x='carat',
    y='price',
    data=diamonds_filtrados
)

plt.show()
```

### Conclusiones de la Actividad 1 — Diamantes

Se analizaron **2.331 diamantes de color F y corte Premium**, seleccionados del dataset `diamonds`, que contiene originalmente 53.940 observaciones.

El coeficiente de correlación de Pearson entre el peso en quilates (`carat`) y el precio (`price`) fue de **r = 0,9131**, lo que indica una **correlación lineal positiva muy fuerte**.

El gráfico de dispersión permitió observar que:

- Los diamantes de mayor peso tienden a presentar precios más elevados.
- La relación presenta una tendencia ascendente, aunque no es perfectamente lineal.
- La dispersión de los precios aumenta entre los diamantes de mayor peso.
- Se identifica una posible observación atípica cercana a los 3 quilates y un precio aproximado de 10.000 dólares.

**Conclusión general:** Existe relación positiva entre el precio y el peso de los diamantes, sin embargo el peso no explica por sí solo toda la diferencias de precios.

Mostrar para cada ubicación, **analítica y gráficamente**, la distribución de **3.a. ejemplares por sexo**, **3.b. ejemplares por especie** y **3.c. desafío: por especie y sexo**. Las tablas siguientes aportan el desglose analítico.

### Consigna 4 — Calcular la matriz de correlación y visualizarla


```python

# Consigna 4 — Calcular la matriz de correlación

# Seleccionar las variables numéricas
penguins_numericos = penguins.select_dtypes(include='number')

# Calcular la matriz de correlación
matriz_penguins = penguins_numericos.corr()

# Mostrar los resultados
matriz_penguins
```

```python

# Consigna 4 — Visualizar la matriz de correlación
sns.heatmap(
    matriz_penguins,
    annot=True,
    cmap='coolwarm',
    vmin=-1,
    vmax=1
)

plt.show()
```

### Consigna 5 — Explorar, interpretar y mostrar gráficamente las variables que tienen mayor y menor correlación

La matriz de correlación de Pearson permitió identificar que la **longitud de las aletas (`flipper_length_mm`) y la masa corporal (`body_mass_g`)** presentan la asociación lineal más fuerte del conjunto de variables analizadas, con un coeficiente de **r = 0,8712**.

Esta correlación positiva muy fuerte indica que los pingüinos con aletas más largas tienden a presentar mayor masa corporal.

El gráfico de dispersión realizado anteriormente confirma visualmente esta tendencia ascendente. Sin embargo, también muestra agrupamientos diferenciados por especie, por lo que las características propias de cada especie contribuyen a la asociación general observada.

**Conclusión:** existe una asociación lineal positiva muy fuerte entre la longitud de las aletas y la masa corporal, aunque esta relación estadística no demuestra causalidad.

```python

# Consigna 5.b — Variables con menor correlación
# Longitud y profundidad del pico

sns.scatterplot(
    data=penguins,
    x='bill_length_mm',
    y='bill_depth_mm',
    hue='species'
)

plt.show()
```

- **Adelie:** `r = 0,4682`, correlación positiva moderada.
- **Chinstrap:** `r = 0,6416`, correlación positiva moderada a fuerte.
- **Gentoo:** `r = 0,7027`, correlación positiva fuerte.

Los resultados muestran que la relación positiva se mantiene dentro de las tres especies, aunque con diferentes intensidades.

**Interpretación:** la correlación general es mayor que las correlaciones individuales, lo que sugiere que las diferencias físicas entre especies contribuyen a fortalecer la asociación observada en el conjunto completo de datos.

Este resultado demuestra la importancia de considerar las variables categóricas al interpretar correlaciones estadísticas.

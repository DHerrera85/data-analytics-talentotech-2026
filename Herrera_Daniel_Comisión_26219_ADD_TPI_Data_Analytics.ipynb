```python
import pandas as pd

#Límpieza de datos

ventas['precio'] = ventas['precio'].astype(str).str.replace(',', '').str.replace('$', '').str.strip()
ventas['cantidad'] = ventas['cantidad'].astype(str).str.strip()

#Convertir el dato

ventas['precio'] = pd.to_numeric(ventas['precio'], errors='coerce')  #Ese coerce intenta convertir los valores a números.
#Si encontrás algo que no puede convertirse, no tires error: transformalo a un NaN
#NaN significa que no tenemos un valor númerico válido.

ventas['cantidad'] = pd.to_numeric(ventas['cantidad'], errors='coerce')

#Eliminar las filas con datos inválidos
ventas = ventas.dropna(subset=['precio', 'cantidad'])

ventas['cantidad'] = ventas['cantidad'].astype(int)

#Columna fecha_venta
ventas['fecha_venta'] = pd.to_datetime(ventas['fecha_venta'], errors='coerce')
ventas = ventas.dropna(subset=['fecha_venta'])   #NaT Not a Time

#Calcular ventas
ventas['total_venta'] = ventas['precio'] * ventas['cantidad']

#Agrupar
#CLASE 7
ventas_mensuales = ventas.groupby(ventas['fecha_venta'].dt.to_period('M'))['total_venta'].sum()

#Mostrar
print("Ventas mensuales:")
display(ventas_mensuales)
```

## 3. Estructura de datos

```python
lista_ventas = []

for indice, fila in ventas.iterrows():
    venta = {}

    venta["producto"] = fila["producto"]
    venta["precio"] = fila["precio"]
    venta["cantidad"] = fila["cantidad"]

    lista_ventas.append(venta)

print("Cantidad de ventas almacenadas:", len(lista_ventas))
```

### Elección de la estructura de datos

# Etapa 2: Preprocesamiento y Limpieza de Datos

## 1. Limpieza de Datos

Se realiza la limpieza del conjunto de ventas, eliminando registros duplicados y preparando los datos para su posterior análisis.

```python
# Cantidad de filas antes de eliminar duplicados
filas_antes = len(ventas)

# Eliminar registros duplicados
ventas = ventas.drop_duplicates()

# Cantidad de filas después de eliminar duplicados
filas_despues = len(ventas)

# Calcular cantidad de duplicados eliminados
duplicados_eliminados = filas_antes - filas_despues

# Mostrar resultados
print("Filas antes de la limpieza:", filas_antes)
print("Filas después de la limpieza:", filas_despues)
print("Duplicados eliminados:", duplicados_eliminados)
```

```python
# Eliminar caracteres no deseados
ventas['precio'] = ventas['precio'].astype(str).str.replace(',', '').str.replace('$', '').str.strip()
ventas['cantidad'] = ventas['cantidad'].astype(str).str.strip()

# Convertir precio y cantidad a valores numéricos
ventas['precio'] = pd.to_numeric(ventas['precio'], errors='coerce')
ventas['cantidad'] = pd.to_numeric(ventas['cantidad'], errors='coerce')

# Eliminar filas con valores inválidos
ventas = ventas.dropna(subset=['precio', 'cantidad'])

# Comprobar el resultado
print("Duplicados restantes:", ventas.duplicated().sum())
print("\nValores nulos restantes:")
print(ventas[['precio', 'cantidad']].isnull().sum())

print("\nTipos de datos:")
print(ventas[['precio', 'cantidad']].dtypes)

print("\nCantidad final de filas:", len(ventas))
```

### Resultado de la limpieza

Se eliminaron 35 registros duplicados del conjunto de ventas. Además, se limpiaron los caracteres no deseados de la columna `precio` y se convirtieron `precio` y `cantidad` a valores numéricos.

Luego de eliminar los registros con valores inválidos, el conjunto quedó conformado por 2998 filas, sin valores nulos en `precio` y `cantidad` y sin registros duplicados.

```python
# Eliminar caracteres no deseados de precio
ventas['precio'] = ventas['precio'].astype(str).str.replace(',', '').str.replace('$', '').str.strip()
```

## 2. Crear una tabla de productos de alto rendimiento

```python
# Calcular el ingreso total de cada venta
ventas['total_venta'] = ventas['precio'] * ventas['cantidad']

# Calcular el promedio de las ventas
promedio_ventas = ventas['total_venta'].mean()

# Filtrar las ventas que superan el promedio
ventas_alto_rendimiento = ventas[ventas['total_venta'] > promedio_ventas]

# Mostrar resultados
print("Promedio de ventas:", promedio_ventas)
print("Cantidad de ventas de alto rendimiento:", len(ventas_alto_rendimiento))

display(ventas_alto_rendimiento.head())
```

## 3. Agregación de Datos

Se agrupan las ventas según la categoría de producto para obtener los ingresos totales generados por cada categoría y analizar su rendimiento.

```python
# Agrupar las ventas por categoría y sumar los ingresos
ventas_por_categoria = ventas.groupby('categoria')['total_venta'].sum()

# Ordenar las categorías de mayor a menor ingreso
ventas_por_categoria = ventas_por_categoria.sort_values(ascending=False)

# Mostrar resultados
print("Ingresos generados por categoría:")
display(ventas_por_categoria)
```

## 4. Integración de Datos

Se integran los conjuntos de ventas y marketing utilizando la columna `producto` como clave común. Esto permite relacionar la información de las ventas con los datos de las campañas de marketing y obtener una visión más amplia de las tendencias.

```python
# Integrar los datos de ventas y marketing mediante la columna producto
ventas_marketing = pd.merge(
    ventas,
    marketing,
    on='producto',
    how='left'
)

# Mostrar información del DataFrame integrado
print("Dimensiones del conjunto integrado:", ventas_marketing.shape)
print("\nPrimeras filas:")
display(ventas_marketing.head())

# Comprobar campañas asociadas
print("\nVentas con información de marketing:")
print(ventas_marketing['id_campanha'].notnull().sum())

print("Ventas sin información de marketing:")
print(ventas_marketing['id_campanha'].isnull().sum())
```

### Resultado de la integración

La integración mediante la columna `producto` permitió relacionar las ventas con las campañas de marketing. Se utilizó una combinación `left` para conservar las ventas y agregar la información de marketing correspondiente.

El conjunto integrado contiene 8994 registros y 12 columnas. Todas las filas resultantes cuentan con información de marketing.

La cantidad de registros es mayor que en el conjunto original de ventas debido a que un mismo producto puede estar asociado a más de una campaña de marketing, generando múltiples coincidencias durante la integración.
